```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
import os 
print(os.getcwd())
```

```python
df=pd.read_csv("../cleaned_data/financial_long.csv")
print(df.head())
```

```python
revenue=df[df["Metric"]=="Total Revenue"]

plt.figure(figsize=(8,5))
plt.plot(revenue["Year"],
         revenue["Value"],
         marker='o')

plt.title("Tesla Total Revenue(2020-2023)")
plt.xlabel("Year")
plt.ylabel("Revenue($)")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout(

)
plt.savefig(
    "../graphs/revenue_trend.png",
    bbox_inches='tight'
)
plt.show()
```
